```python
from particles_mod.Icosphere import *
import hydrodynamic_int.hessian as hess
import numpy as np
import pyvista as pv
import matplotlib.pyplot as plt
from scipy.spatial.transform import Rotation as R
np.set_printoptions(precision=3)
pv.set_jupyter_backend('html')
```

```python

radius = 3.0
density = 2.0
Kpair = 20
Kdi = 0.5

icosphere = IcoSphere(radius, density, Kpair, Kdi)
hessian = icosphere.calculate_hessian()

    
```

```python



positions = np.array(icosphere.get_positions())
angles = np.linspace(0, 2*np.pi, 50)
energies = []
for angle in angles:
    rotation_matrix = R.from_euler('z', angle).as_matrix()
    rotated_positions = np.array([rotation_matrix @ pos for pos in positions])
    rotation_displacement = rotated_positions - positions
    rotation_displacement = rotation_displacement.flatten()
    rotation_energy = rotation_displacement.T @ hessian @ rotation_displacement
    energies.append(rotation_energy)

angles = np.array(angles)
energies = np.array(energies)

plt.figure(figsize=(6, 4))
plt.polar(angles, energies)
plt.title('Rotation Energy vs Angle')
plt.grid(True)
number_of_ticks = 4
plt.yticks(np.linspace(0, max(energies), number_of_ticks),
              [f'{int(e)}' for e in np.linspace(0, max(energies), number_of_ticks)])
plt.show()


```
