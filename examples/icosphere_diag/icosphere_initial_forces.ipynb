```python
from particles_mod import IcoSphere as ico
from hydrodynamic_int.hessian import obtainHessian
import numpy as np
import pandas as pd
import pyvista as pv
pv.set_jupyter_backend('html')
```

```python
r = 2
dens = 2
Kpair = 10.5
Kdi = 2.5


icosphere = ico(radius=r, density=dens, Kpair=Kpair, Kdi=Kdi)
```

```python


bonds = icosphere.bonds

# print bond keys
print("Bond keys:")
for key in bonds.keys():
    for k in bonds[key].keys():
        print(f"{key} - {k}")

for key in bonds:
    print(f"{key} labels:")
    for i in range(len(bonds[key]["labels"])):
        print(f"Bond {i} label: {bonds[key]['labels'][i]}")
    print(f"{key} data:")
    for i in range(len(bonds[key]["data"])):
        print(f"Bond {i} data: {bonds[key]['data'][i]}")

```

```python

def read_forces_file(filename):
    forces = pd.read_csv(filename, delim_whitespace=True, index_col=0)
    forces = forces[['Fx', 'Fy', 'Fz']].values
    forces = np.array(forces)
    return forces

forces = read_forces_file("results/forces.txt")
print(forces.shape)

positions = np.array(positions)

print(positions.shape)


def create_forces_plot(positions, forces):
    plotter = pv.Plotter()
    mesh = pv.PolyData(positions)
    mesh = mesh.delaunay_3d()
    plotter.add_mesh(mesh, color='white', point_size=5, render_points_as_spheres=True)
    plotter.add_arrows(positions, forces)
    return plotter

forces_plot = create_forces_plot(positions, forces)
forces_plot.show()




```
