```python
import sympy as sp
from particles_mod.PlainMembrane import *
import numpy as np

membrane = Plain_Membrane(Density = 0.8)

k_spring = sp.symbols('k')
l0_equilibrium = sp.symbols('l0')
positions = [sp.Matrix([sp.symbols(f'{a}_{i}') for a in ["x", "y", "z"]]) for i in range(2)]


def obtain_distance(pos1, pos2):
    delta_r = pos1 - pos2
    return sp.sqrt(delta_r.dot(delta_r))

energy = 0.5 * k_spring * (obtain_distance(positions[0], positions[1]) - l0_equilibrium)**2

variables = sp.Matrix([a for pos in positions for a in pos])

print("Variables:")
sp.pprint(variables)

hessian = energy.diff(variables).jacobian(variables)

sp.pprint(energy)




```
