```python
import sympy as sp
from particles_mod.PlainMembrane import *
import numpy as np

membrane = Plain_Membrane(Density = 0.8)

positions = membrane.generate_positions()
bonds = membrane.generate_bonds()

k_spring = sp.symbols('k')
l0_equilibrium = sp.symbols('l0')



def create_symbolic_positions():
    nparticles = 2
    positions = np.array([sp.Matrix([0,0,0]) for i in range(nparticles)])
    for i in range(nparticles):
        positions[i][0] = -0.5 + i/(nparticles-1)
    return positions

positions = create_symbolic_positions()

def energy_spring(pos, k, l0):
    """
    Calculate the energy of a spring between two particles.
    :param pos: symbolic positions of the particles
    :param k: spring constant
    :param l0: equilibrium length
    """
    # calculate the distance between the two particles
    r = sp.sqrt((pos[0][0] - pos[1][0])**2 + (pos[0][1] - pos[1][1])**2 + (pos[0][2] - pos[1][2])**2)
    # calculate the energy of the spring
    energy = 0.5 * k * (r - l0)**2
    return energy

print("Symbolic energy of a spring between two particles:")
energy = energy_spring(positions, k_spring, l0_equilibrium)
print(energy) 


```
